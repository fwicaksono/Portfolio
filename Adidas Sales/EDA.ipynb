```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("Adidas US Sales Datasets.xlsx")
```

```python
df
```

```python
# Remove first 3 rows
df = df.iloc[3:]
df.columns = df.iloc[0]  #make the first row as column names
df = df[1:].reset_index(drop=True) #remove header duplicate
df = df.iloc[:, 1:] # remove first column where contains all NaN 
df = df.reset_index(drop=True)
df = df.copy()  # Buat salinan agar tidak terpengaruh oleh cache
df = df.reset_index(drop=True)  # Hapus index lama
df.index.name = None  # Hapus nama index
df.columns.name = None  # Hapus nama kolom kalau ada
df = df.rename_axis(None, axis=1)  # Paksa hapus nama index kolom
df
```

```python
print(df.isnull().sum())   #Cek NULL setiap kolom
```

```python
print(df.isnull().values.any())   #cek NULL di seluruh DF
```
